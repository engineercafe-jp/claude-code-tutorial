# 4️⃣ Essential Commands for Beginners (1/2)

## ⌨️ Keyboard Shortcuts

### 🔄 Shift + Tab — Switch Modes

Each press cycles through modes:

```
Normal Mode → Auto-Accept Mode → Plan Mode → Normal Mode ...
```

| Mode | Indicator | Description |
|------|-----------|-------------|
| **Normal** | (default) | Standard mode — confirms before making changes |
| **Auto-Accept** | `⏵⏵ accept edits on` | Approves changes automatically |
| **Plan** | `⏸ plan mode on` | **Read-only** — planning only, no edits |

---

**💡 Plan Mode is especially important!** → Details on the later slide

## ⌨️ Slash Commands (2/2)

These commands are available at any time during a session:

| Command | Description | When to use |
|---------|-------------|-------------|
| `/model` | Change the AI model | Adjust speed vs. accuracy |
| `/status` | Check current status | See what's going on |
| `/resume` | Resume a previous session | Continue yesterday's work |
| `/clear` | Reset conversation history | Switch to a different task |
| `/init` | Auto-generate CLAUDE.md | When starting a new project |

---

### Other Shortcuts

| Shortcut | Description |
|----------|-------------|
| `?` | Show keyboard shortcut list |
| `Esc` | Stop the current action |
| `Ctrl+C` | Exit Claude Code |

# 🗺️ What is Plan Mode?

## Switch with Shift+Tab — Planning-Only Mode

**Plan Mode** = Claude reads files **without making any edits** 🔍

---

## 🎯 Why use it?

> **"Plan before you implement!"**

Jumping straight into complex changes can cause:
- ❌ Unexpected files getting modified
- ❌ Too much code written to review
- ❌ Hard-to-undo mistakes

With Plan Mode:
- ✅ Claude only reads and analyzes
- ✅ You review the plan before anything is changed
- ✅ Implementation starts only after you approve

---

## 📋 Recommended Workflow

```
1. Explore & plan in Plan Mode 🔍
2. Review & approve the plan ✅
3. Implement in Normal Mode 💻
4. Test & commit 🎉
```

# 6️⃣ Practice! Get Fukuoka City Weather Forecast 📋

We'll fetch real-time weather forecast data from the Japan Meteorological Agency API.

---

Tell Claude Code:

```
Using the JMA API, create a Python script that fetches
and displays the weather forecast for Fukuoka City
```

````python
import urllib.request
import json

# Japan Meteorological Agency API: Fukuoka Prefecture (area code 400000)
url = "https://www.jma.go.jp/bosai/forecast/data/forecast/400000.json"

with urllib.request.urlopen(url) as response:
    data = json.loads(response.read())

# Get forecast for Fukuoka area
area_name = data[0]["timeSeries"][0]["areas"][0]["area"]["name"]
weather_codes = data[0]["timeSeries"][0]["areas"][0]["weatherCodes"]
time_defines = data[0]["timeSeries"][0]["timeDefines"]

# Weather code → description
weather_map = {
    "100": "☀️ Sunny",
    "101": "🌤️ Partly Cloudy",
    "200": "☁️ Cloudy",
    "201": "🌥️ Mostly Cloudy",
    "300": "🌧️ Rainy",
    "400": "❄️ Snowy",
}

print(f"📍 Weather Forecast: {area_name}")
print("-" * 30)
for time, code in zip(time_defines[:3], weather_codes[:3]):
    date = time[:10]
    weather = weather_map.get(code, f"Code: {code}")
    print(f"{date}: {weather}")
````

# 📎 Appendix: Upload to GitHub

Save and share your files on **GitHub** 🌐

## What is GitHub?

- A service for storing and publishing code online
- Used by developers worldwide
- Free to get started

---

First, create an account at [github.com](https://github.com)!

## 1️⃣ Install & Authenticate GitHub CLI (gh)

**GitHub CLI** lets you manage GitHub entirely from the command line.

---

### Install

**🍎 macOS**
```bash
brew install gh
```

**🪟 Windows (PowerShell)**
```powershell
winget install --id GitHub.cli
```

---

### Authenticate (first time only)

```bash
gh auth login
```

> Follow the interactive prompts. Choose `GitHub.com` → `HTTPS` → `Login with a web browser` for the easiest setup 🎉

## 2️⃣ Create a Repository and Upload

Move into your working folder, then run a single command:

---

```bash
cd hello_world
gh repo create hello-world --public --source=. --push
```

| Option | Description |
|--------|-------------|
| `--public` | Create a public repository |
| `--source=.` | Use the current folder as the source |
| `--push` | Upload files immediately after creation |

---

**🎉 One command — repository created and files uploaded!**

## 3️⃣ Add or Update Files

For subsequent uploads, use git commands:

```bash

git push
```

---

### View your repository in the browser

```bash

```

> Opens your repository page in the browser 🌐
