# 2️⃣ Claude Code のインストール

## 💻 GUI版（デスクトップアプリ）

### Mac / Windows

1. 公式サイト [claude.ai/code](https://claude.ai/code) からダウンロード
2. インストーラーを実行
3. Anthropic アカウントでログイン

---

**🎯 一番かんたん！コマンド操作が不要**

## 🤖 CLI版（コマンドライン）

依存関係なし・自動更新つきの**ネイティブインストーラー**が推奨です 🎯

### 🍎 macOS / 🐧 Linux

```bash
curl -fsSL https://claude.ai/install.sh | bash
```

### 🪟 Windows（PowerShell）

```powershell
irm https://claude.ai/install.ps1 | iex
```

> **前提:** [Git for Windows](https://git-scm.com/downloads/win) をインストールしておく

### 起動方法（共通）

```bash
claude
```

---

**💡 ヒント:** 初回起動時に Anthropic アカウントへのログインが求められます

# 4️⃣ 初心者が抑えるべきコマンド

## ⌨️ キーボードショートカット

### 🔄 Shift + Tab ― モードの切り替え

押すたびにモードが切り替わります：

```
Normal Mode → Auto-Accept Mode → Plan Mode → Normal Mode ...
```

| モード | アイコン | 説明 |
|--------|---------|------|
| **Normal** | （デフォルト） | 通常モード。変更前に確認を求められる |
| **Auto-Accept** | `⏵⏵ accept edits on` | 変更を自動で承認。スピード重視 |
| **Plan** | `⏸ plan mode on` | **読み取り専用**。計画だけ立てる |

---

**💡 Plan Mode が特に重要！**

# 🗺️ Plan Mode とは？

## Shift+Tab で切り替える「計画専用モード」

**Plan Mode** = Claude がファイルを**編集しないで**、読むだけのモード 🔍

---

## 🎯 なぜ使うの？

> **「実装する前に計画を立てよう！」**

複雑な変更をいきなり実装すると…
- ❌ 意図しないファイルが変わる
- ❌ 大量のコードが書かれて把握できない
- ❌ やり直しが大変

Plan Mode を使うと…
- ✅ コードを読んで分析するだけ
- ✅ 「こう実装します」という計画を先に確認できる
- ✅ 計画に同意してから実装スタート

---

## 📋 推奨ワークフロー

```
1. Plan Mode で探索・計画 🔍
2. 計画を確認・承認 ✅
3. Normal Mode で実装 💻
4. 動作確認・コミット 🎉
```

## ⌨️ スラッシュコマンド

セッション中にいつでも使えるコマンドです：

| コマンド | 説明 | いつ使う？ |
|---------|------|----------|
| `/model` | 使用モデルを変更 | 速度 vs 精度を調整したい時 |
| `/status` | ステータスを確認 | 現在の状況を把握したい時 |
| `/resume` | 過去のセッションを再開 | 昨日の作業を続けたい時 |
| `/clear` | 会話履歴をリセット | 別のタスクに切り替える時 |
| `/init` | CLAUDE.md を自動生成 | プロジェクト開始時に |

---

### その他のショートカット

| ショートカット | 説明 |
|-------------|------|
| `?` | キーボードショートカット一覧を表示 |
| `Esc` | 実行中のアクションを停止 |
| `Ctrl+C` | Claude Code を終了 |

## GitHub とは？

- コードをインターネット上に保存・公開できるサービス
- 世界中の開発者が使うプラットフォーム
- 無料アカウントで始められる

---

まずは [github.com](https://github.com) でアカウントを作成しておこう！

## 1️⃣ GitHub CLI（gh）をインストール・認証する

**GitHub CLI** を使うと、コマンドだけで GitHub 操作が完結します。

---

### インストール

**🍎 macOS**
```bash
brew install gh
```

**🪟 Windows（PowerShell）**
```powershell
winget install --id GitHub.cli
```

---

### 認証（初回のみ）

```bash
gh auth login
```

> 対話形式で進む。`GitHub.com` → `HTTPS` → `Login with a web browser` を選ぶと簡単 🎉

## 2️⃣ リポジトリを作成してアップロードする

作業フォルダに移動してから、以下のコマンドを実行するだけ！

---

```bash
cd hello_world
gh repo create
```

---

**🎉 1コマンドでリポジトリ作成 → アップロードまで完了！**
