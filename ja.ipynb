# 2️⃣ Claude Code のインストール

### Mac / Windows

```

```

> **前提:** [Git for Windows](https://git-scm.com/downloads/win) をインストールしておく

### 起動方法（共通）

```bash
claude
```

**💡 ヒント:** 初回起動時に Anthropic アカウントへのログインが求められます

# 4️⃣ 初心者が抑えるべきコマンド (1/2)

## ⌨️ キーボードショートカット

### 🔄 Shift + Tab ― モードの切り替え

押すたびにモードが切り替わります：

```
Normal Mode → Auto-Accept Mode → Plan Mode → Normal Mode ...
```

| モード | アイコン | 説明 |
|--------|---------|------|
| **Normal** | （デフォルト） | 通常モード。変更前に確認を求められる |
| **Auto-Accept** | `⏵⏵ accept edits on` | 変更を自動で承認。スピード重視 |
| **Plan** | `⏸ plan mode on` | **読み取り専用**。計画だけ立てる |

---

**💡 Plan Mode が特に重要！** → 別のスライドで詳しく説明

## ⌨️ スラッシュコマンド (2/2)

セッション中にいつでも使えるコマンドです：

| コマンド | 説明 | いつ使う？ |
|---------|------|----------|
| `/model` | 使用モデルを変更 | 速度 vs 精度を調整したい時 |
| `/status` | ステータスを確認 | 現在の状況を把握したい時 |
| `/resume` | 過去のセッションを再開 | 昨日の作業を続けたい時 |
| `/clear` | 会話履歴をリセット | 別のタスクに切り替える時 |
| `/init` | CLAUDE.md を自動生成 | プロジェクト開始時に |

---

### その他のショートカット

| ショートカット | 説明 |
|-------------|------|
| `?` | キーボードショートカット一覧を表示 |
| `Esc` | 実行中のアクションを停止 |
| `Ctrl+C` | Claude Code を終了 |

# 🗺️ Plan Mode とは？

## Shift+Tab で切り替える「計画専用モード」

**Plan Mode** = Claude がファイルを**編集しないで**、読むだけのモード 🔍

---

## 🎯 なぜ使うの？

> **「実装する前に計画を立てよう！」**

複雑な変更をいきなり実装すると…
- ❌ 意図しないファイルが変わる
- ❌ 大量のコードが書かれて把握できない
- ❌ やり直しが大変

Plan Mode を使うと…
- ✅ コードを読んで分析するだけ
- ✅ 「こう実装します」という計画を先に確認できる
- ✅ 計画に同意してから実装スタート

---

## 📋 推奨ワークフロー

```
1. Plan Mode で探索・計画 🔍
2. 計画を確認・承認 ✅
3. Normal Mode で実装 💻
4. 動作確認・コミット 🎉
```

# 6️⃣ 実践！福岡市の天気予報を取得しよう 📋

気象庁の API からリアルタイムの天気予報データを取得します。

---

Claude Code にこう伝えよう：

```
気象庁のAPIを使って福岡市の天気予報を取得して表示する
Pythonスクリプトを作って
```

````python
import urllib.request
import json

# 気象庁API: 福岡県（エリアコード 400000）
url = "https://www.jma.go.jp/bosai/forecast/data/forecast/400000.json"

with urllib.request.urlopen(url) as response:
    data = json.loads(response.read())

# 福岡市の予報エリアを取得
area_name = data[0]["timeSeries"][0]["areas"][0]["area"]["name"]
weather_codes = data[0]["timeSeries"][0]["areas"][0]["weatherCodes"]
time_defines = data[0]["timeSeries"][0]["timeDefines"]

# 天気コード → 天気の説明
weather_map = {
    "100": "☀️ 晴れ",
    "101": "🌤️ 晴れ時々曇り",
    "200": "☁️ 曇り",
    "201": "🌥️ 曇り時々晴れ",
    "300": "🌧️ 雨",
    "400": "❄️ 雪",
}

print(f"📍 {area_name} の天気予報")
print("-" * 30)
for time, code in zip(time_defines[:3], weather_codes[:3]):
    date = time[:10]
    weather = weather_map.get(code, f"天気コード: {code}")
    print(f"{date}: {weather}")
````

## GitHub とは？

- コードをインターネット上に保存・公開できるサービス
- 世界中の開発者が使うプラットフォーム
- 無料アカウントで始められる

まずは [github.com](https://github.com) でアカウントを作成しておこう！

## 1️⃣ リポジトリを作成する

**リポジトリ** = ファイルをまとめて管理するフォルダのようなもの

---

### 手順

1. [github.com](https://github.com) にログイン
2. 右上の **`+`** ボタン → **New repository** をクリック
3. 以下を入力して **Create repository** をクリック

| 項目 | 入力例 |
|------|--------|
| Repository name | `hello-world` |
| Description | 任意（省略可） |
| Public / Private | **Public**（公開）または Private（非公開） |
| Add a README file | ✅ チェックを入れる |

---

**🎉 リポジトリの完成！**

## 2️⃣ ファイルをアップロードする

ブラウザからコピー＆ペースト感覚でアップロードできます！

---

### 手順

1. 作成したリポジトリのページを開く
2. **Add file** → **Upload files** をクリック
3. ファイルをドラッグ＆ドロップ、または **choose your files** からファイルを選択
4. 下部の **Commit changes** をクリック

---

### 💡 ヒント

- 複数ファイルを一度にアップロードできる
- フォルダごとドラッグ＆ドロップも可能

---

**✅ これでファイルが GitHub 上に公開されました！**
